```python

import socket
import os
import threading

HOST = "127.0.0.1"
PORT = 5005

def start_server():
    server_socket = socket.socket(socket.AF_INET, socket.SOCK_DGRAM)
    server_socket.bind((HOST, PORT))

    print(f"UDP Server listening on {HOST}:{PORT}")

    while True:
        filename, client_addr = server_socket.recvfrom(1024)
        filename = filename.decode().strip()

        print(f"Client requested file: {filename}")

        if os.path.isfile(filename):
            with open(filename, "r") as f:
                data = f.read()

            server_socket.sendto(data.encode(), client_addr)
        else:
            server_socket.sendto(
                b"ERROR: File not found", client_addr
            )

# Run server in background
server_thread = threading.Thread(
    target=start_server,
    daemon=True
)
server_thread.start()
```

```python

import socket

HOST = "127.0.0.1"
PORT = 5005

client_socket = socket.socket(
    socket.AF_INET,
    socket.SOCK_DGRAM
)

client_socket.settimeout(5)

filename = input("Enter filename to request: ")

client_socket.sendto(
    filename.encode(),
    (HOST, PORT)
)

try:
    data, server_addr = client_socket.recvfrom(65507)

    print("\n--- File Content ---")
    print(data.decode())

except socket.timeout:
    print("Server did not respond.")

finally:
    client_socket.close()
```
